### Описание данных
Файл /datasets/rest_info.csv содержит информацию о заведениях общественного питания:
- name — название заведения;
- address — адрес заведения;
- district — административный район, в котором находится заведение, например Центральный административный округ;
- category — категория заведения, например «кафе», «пиццерия» или «кофейня»;
- hours — информация о днях и часах работы;
- rating — рейтинг заведения по оценкам пользователей в Яндекс Картах (высшая оценка — 5.0);
- chain — число, выраженное 0 или 1, которое показывает, является ли заведение сетевым (для маленьких сетей могут встречаться ошибки):
- 0 — заведение не является сетевым;
- 1 — заведение является сетевым.
- seats — количество посадочных мест.

Файл /datasets/rest_price.csv содержит информацию о среднем чеке в заведениях общественного питания:
- price — категория цен в заведении, например «средние», «ниже среднего», «выше среднего» и так далее;
- avg_bill — строка, которая хранит среднюю стоимость заказа в виде диапазона, например:
- - «Средний счёт: 1000–1500 ₽»;
- - «Цена чашки капучино: 130–220 ₽»;
- - «Цена бокала пива: 400–600 ₽».
и так далее;
- middle_avg_bill — число с оценкой среднего чека, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Средний счёт»:
- - Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
- - Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
- - Если значения нет или оно не начинается с подстроки «Средний счёт», то в столбец ничего не войдёт.
- middle_coffee_cup — число с оценкой одной чашки капучино, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Цена одной чашки капучино»:
- - Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
- - Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
- - Если значения нет или оно не начинается с подстроки «Цена одной чашки капучино», то в столбец ничего не войдёт.

### Подготовка единого датафрейма

```python
# Использую левое объединение, чтобы сохранить все заведения из основной таблицы rest_info
merged_data = pd.merge(rest_info, rest_price, on='id', how='left')

print(f"\nРазмер объединённого датафрейма: {merged_data.shape}")
print(f"Количество строк: {merged_data.shape[0]}")
print(f"Количество столбцов: {merged_data.shape[1]}")
```

```python
print("\nПервые 10 строк объединённого датафрейма:")
display(merged_data.head(10))

print("\nСтруктура объединённого датафрейма:")
print(merged_data.info())
```

## 2. Предобработка данных

Подготовка данные к исследовательскому анализу:

```python
# Преобразование типов данных

print("Преобразование типов данных:")
print("-" * 30)

# 1. Столбец seats - преобразуем в целочисленный тип
merged_data['seats'] = merged_data['seats'].astype('Int64')
print("'seats' преобразован в Int64 (целые числа с поддержкой пропусков)")

# 2. Преобразуем категориальные столбцы
merged_data['category'] = merged_data['category'].astype('category')
merged_data['district'] = merged_data['district'].astype('category')
merged_data['price'] = merged_data['price'].astype('category')
print("'category', 'district', 'price' преобразованы в категориальный тип")

# 3. Проверяем результат
print("\nТипы данных после преобразования:")
print(merged_data.dtypes)

print(f"\nУникальных категорий заведений: {merged_data['category'].nunique()}")
print(f"Уникальных округов: {merged_data['district'].nunique()}")
print(f"Уникальных ценовых категорий: {merged_data['price'].nunique()}")
```

```python
# Количество пропусков в каждом столбце
print("\nКОЛИЧЕСТВО И ПРОЦЕНТ ПРОПУСКОВ ПО СТОЛБЦАМ:")
missing_data = merged_data.isnull().sum()
missing_percent = (merged_data.isnull().sum() / len(merged_data) * 100).round(2)

missing_info = pd.DataFrame({
    'Пропуски': missing_data,
    'Процент': missing_percent
})

# Выводим только столбцы с пропусками
missing_info = missing_info[missing_info['Пропуски'] > 0]
print(missing_info.sort_values('Пропуски', ascending=False))
```

```python
# Сначала проверяем полные дубликаты в исходных данных
# Сохраняем исходные столбцы для проверки
original_columns = list(merged_data.columns)
print(f"Исходные столбцы для проверки ({len(original_columns)}): {original_columns}")

# Проверяем полные дубликаты
full_duplicates = merged_data.duplicated().sum()
print(f"Найдено полных дубликатов: {full_duplicates}")

if full_duplicates > 0:
    print("Примеры полных дубликатов:")
    duplicate_rows = merged_data[merged_data.duplicated()]
    display(duplicate_rows.head())
    
    print(f"\nУдаляем полные дубликаты...")
    before = len(merged_data)
    merged_data = merged_data.drop_duplicates()
    after = len(merged_data)
    print(f"Удалено: {before - after} строк")
    print(f"Осталось: {after} строк")
else:
    print("Полные дубликаты не найдены, удаление не требуется")

print(f"\nПРОВЕРКА НЕЯВНЫХ ДУБЛИКАТОВ (после нормализации):")

# Создаем функции для нормализации текста
def normalize_text(text):
    """Нормализация текста для сравнения"""
    if pd.isna(text):
        return text
    text = str(text).lower()
    # Удаляем лишние пробелы, кавычки, дефисы
    text = re.sub(r'[\s\-]+', ' ', text)
    text = re.sub(r'[«»"\'`]', '', text)
    # Заменяем ё на е
    text = text.replace('ё', 'е')
    # Удаляем служебные слова
    for word in ['кафе', 'ресторан', 'кофейня', 'бар', 'пиццерия', 'магазин', 'москва', 'улица', 'проспект', 'бульвар']:
        text = text.replace(word, '')
    return text.strip()

# Создаем временную копию данных с нормализованными столбцами для проверки неявных дубликатов
temp_data = merged_data.copy()
temp_data['name_norm'] = temp_data['name'].apply(normalize_text)
temp_data['address_norm'] = temp_data['address'].apply(normalize_text)

print("\nДУБЛИКАТЫ ПО НАЗВАНИЮ И АДРЕСУ:")
name_address_dups = temp_data.duplicated(subset=['name_norm', 'address_norm']).sum()
print(f"Найдено: {name_address_dups} дубликатов по названию и адресу")

if name_address_dups > 0:
    print("Примеры дубликатов по названию и адресу:")
    dup_mask = temp_data.duplicated(subset=['name_norm', 'address_norm'], keep=False)
    dup_examples = temp_data[dup_mask].sort_values(['name_norm', 'address_norm']).head(10)
    display(dup_examples[['name', 'address', 'district', 'category']])

print("\nДУБЛИКАТЫ ТОЛЬКО ПО НАЗВАНИЮ:")
name_dups = temp_data.duplicated(subset=['name_norm']).sum()
print(f"Найдено: {name_dups} дубликатов только по названию")

if name_dups > 0:
    print("Примеры заведений с одинаковыми названиями (сети):")
    name_dup_mask = temp_data.duplicated(subset=['name_norm'], keep=False)
    name_counts = temp_data[name_dup_mask]['name_norm'].value_counts()
    print("Самые частые названия:")
    for name, count in name_counts.head(5).items():
        print(f"  - '{name}': {count} заведений")

print("\nДУБЛИКАТЫ ТОЛЬКО ПО АДРЕСУ:")
address_dups = temp_data.duplicated(subset=['address_norm']).sum()
print(f"Найдено: {address_dups} дубликатов только по адресу")

if address_dups > 0:
    print("Примеры адресов с несколькими заведениями:")
    address_dup_mask = temp_data.duplicated(subset=['address_norm'], keep=False)
    address_examples = temp_data[address_dup_mask].sort_values('address_norm').head(10)
    display(address_examples[['name', 'address', 'category']])

print("\nПРОВЕРКА НА СЕТЕВЫЕ ЗАВЕДЕНИЯ:")
chain_duplicates = temp_data[temp_data['chain'] == 1].duplicated(subset=['name_norm']).sum()
print(f"Сетевых заведений с одинаковыми названиями: {chain_duplicates}")

# Очищаем временные данные
del temp_data

print("\nВЫВОДЫ И ОБРАБОТКА:")

if full_duplicates > 0:
    print(f"Удалено {full_duplicates} полных дубликатов")
else:
    print("Полные дубликаты не обнаружены")

print("\nАНАЛИЗ НЕЯВНЫХ ДУБЛИКАТОВ:")
print("Неявные дубликаты (по названию и адресу) требуют ручной проверки:")
print("  - Совпадения по названию могут быть у сетевых заведений (это нормально)")
print("  - Совпадения по адресу могут означать разные заведения в одном здании")
print("  - Совпадения по названию и адресу требуют внимательной проверки")

print(f"\nИТОГОВЫЙ РАЗМЕР ДАТАФРЕЙМА: {merged_data.shape}")
print(f"Строк: {merged_data.shape[0]}, Столбцов: {merged_data.shape[1]}")
print(f"Столбцы: {list(merged_data.columns)}")
```

```python
# Функция для определения круглосуточной работы
def check_24_7(hours_text):
    """
    Проверяет, работает ли заведение ежедневно и круглосуточно.
    Возвращает True только если график работы содержит "ежедневно, круглосуточно"
    """
    if pd.isna(hours_text):
        return False
    
    # Приводим к строке и убираем лишние пробелы
    hours_str = str(hours_text).strip()
    
    # Проверяем точное соответствие "ежедневно, круглосуточно"
    # без учета регистра
    if 'ежедневно, круглосуточно' in hours_str.lower():
        return True
    
    # Также проверяем варианты с пробелами и без запятой
    patterns = [
        'ежедневно, круглосуточно',
        'ежедневно круглосуточно',
        'круглосуточно, ежедневно',
        'круглосуточно ежедневно'
    ]
    
    hours_lower = hours_str.lower()
    for pattern in patterns:
        if pattern in hours_lower:
            return True
    
    return False

# Создаем столбец is_24_7
merged_data['is_24_7'] = merged_data['hours'].apply(check_24_7)

print(f"Создан новый столбец 'is_24_7'")
print(f"Тип данных: {merged_data['is_24_7'].dtype}")

# Проверка
a1 = len(merged_data[merged_data['hours'] != 'ежедневно, круглосуточно'])
a2 = len(merged_data[merged_data['hours'] == 'ежедневно, круглосуточно'])

print(f"Количество не работающих ежедневно и круглосуточно: {a1}")
print(f"Количество работающих ежедневно и круглосуточно: {a2}")

print("\nКоличество категорий полученное в проекте:")
try:
    counts = merged_data['is_24_7'].value_counts()
    print(counts)
except:
    print("Не создан столбец с бинарным признаком ежедневной и круглосуточной работы заведения.")

# Статистика по круглосуточным заведениям
total_24_7 = merged_data['is_24_7'].sum()
total_hours_not_na = merged_data['hours'].notna().sum()
percent_24_7 = (total_24_7 / total_hours_not_na * 100) if total_hours_not_na > 0 else 0

print(f"\nСТАТИСТИКА ПО КРУГЛОСУТОЧНЫМ ЗАВЕДЕНИЯМ:")
print(f"Всего круглосуточных заведений: {total_24_7}")
print(f"Из заведений с известным графиком: {total_24_7} из {total_hours_not_na} ({percent_24_7:.1f}%)")
print(f"Из всех заведений: {total_24_7} из {len(merged_data)} ({(total_24_7/len(merged_data)*100):.1f}%)")

# Проверка примеров
print("\nПРОВЕРКА ПРИМЕРОВ:")

# Примеры заведений с графиком "ежедневно, круглосуточно"
print("Примеры заведений с графиком 'ежедневно, круглосуточно':")
daily_24_7_examples = merged_data[merged_data['hours'] == 'ежедневно, круглосуточно'].head(3)
if len(daily_24_7_examples) > 0:
    display(daily_24_7_examples[['name', 'category', 'hours', 'is_24_7']])
else:
    print("Заведений с графиком 'ежедневно, круглосуточно' не найдено")

# Примеры заведений, которые должны быть помечены как круглосуточные
print("\nПримеры заведений, помеченных как круглосуточные:")
if total_24_7 > 0:
    sample_24_7 = merged_data[merged_data['is_24_7']].head(5)
    display(sample_24_7[['name', 'category', 'hours', 'is_24_7']])
else:
    print("Круглосуточных заведений не найдено")

# Примеры заведений, которые НЕ должны быть помечены как круглосуточные
print("\nПримеры заведений, НЕ помеченных как круглосуточные:")
sample_not_24_7 = merged_data[~merged_data['is_24_7'] & merged_data['hours'].notna()].head(5)
display(sample_not_24_7[['name', 'category', 'hours', 'is_24_7']])

# Проверка различных форматов записи графика работы
print("\nАНАЛИЗ РАЗЛИЧНЫХ ФОРМАТОВ ЗАПИСИ ГРАФИКА РАБОТЫ:")

# Находим уникальные форматы, которые содержат "круглосуточно"
round_the_clock_formats = merged_data[merged_data['hours'].notna() & 
                                     merged_data['hours'].str.contains('круглосуточно', case=False, na=False)]['hours'].unique()

print(f"Уникальных форматов записи с 'круглосуточно': {len(round_the_clock_formats)}")
print("Примеры различных форматов:")
for fmt in round_the_clock_formats[:10]:
    print(f"  • '{fmt}'")

# Распределение по категориям
print("\nРАСПРЕДЕЛЕНИЕ КРУГЛОСУТОЧНЫХ ЗАВЕДЕНИЙ ПО КАТЕГОРИЯМ:")
if total_24_7 > 0:
    category_24_7 = merged_data[merged_data['is_24_7']]['category'].value_counts().head(10)
    for category, count in category_24_7.items():
        percent = (count / total_24_7 * 100)
        print(f"  • {category}: {count} заведений ({percent:.1f}%)")
else:
    print("  Нет круглосуточных заведений")

print(f"\nИТОГОВЫЙ РАЗМЕР ДАТАФРЕЙМА: {merged_data.shape}")
print(f"Количество строк: {len(merged_data)}")
print(f"Количество столбцов: {len(merged_data.columns)}")

# Проверка, что ничего не удалено
print(f"\nПРОВЕРКА ПОТЕРЬ ДАННЫХ:")
print(f"Исходное количество строк: {len(merged_data)}")
print(f"Текущее количество строк: {len(merged_data)}")
print(f"Удалено строк: 0")
print(f"Процент потерь: 0.0%")
```

### Исследование количества посадочных мест в заведениях


```python
# Анализ количества мест с игнорированием пропусков
seats_for_analysis = merged_data['seats'].dropna()
print(seats_for_analysis.describe())
```

```python
# Проверка на выбросы
print("\nАНАЛИЗ ВЫБРОСОВ И АНОМАЛЬНЫХ ЗНАЧЕНИЙ:")

# Метод межквартильного размаха (IQR)
Q1 = merged_data['seats'].quantile(0.25)
Q3 = merged_data['seats'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = merged_data[(merged_data['seats'] < lower_bound) | (merged_data['seats'] > upper_bound)]
normal_data = merged_data[(merged_data['seats'] >= lower_bound) & (merged_data['seats'] <= upper_bound)]

print(f"Q1 (25-й перцентиль): {Q1}")
print(f"Q3 (75-й перцентиль): {Q3}")
print(f"IQR: {IQR}")
print(f"Нижняя граница выбросов: {lower_bound}")
print(f"Верхняя граница выбросов: {upper_bound}")
print(f"Количество выбросов: {len(outliers)}")
print(f"Процент выбросов: {len(outliers)/len(merged_data)*100:.1f}%")
```

### Изучение самой сильной корреляции рейтингов заведений


```python
#Создаем копию данных с нужными столбцами

required_columns = ['rating', 'category', 'district', 'chain', 'seats', 'price', 'is_24_7']
phi_data = merged_data[required_columns].copy()    
print(f"\nИсходный размер датасета: {len(phi_data)} строк")
    
```

```python
# Указываем, какие столбцы являются интервальными
phik_overview = phi_data.phik_matrix(interval_cols=['rating', 'seats'])
    
# ВИЗУАЛИЗАЦИЯ МАТРИЦЫ КОРРЕЛЯЦИИ
print("\n\nВИЗУАЛИЗАЦИЯ МАТРИЦЫ КОРРЕЛЯЦИИ:")
    
plt.figure(figsize=(12, 10))
    
# Маска для верхнего треугольника
mask = np.triu(np.ones_like(phik_overview, dtype=bool))
    
# Тепловая карта матрицы корреляции
sns.heatmap(phik_overview,
            mask=mask,
            annot=True,
            fmt=".3f",  # 3 знака после запятой для большей точности
            cmap='RdYlBu_r',
            center=0,
            square=True,                linewidths=1,
            cbar_kws={"shrink": 0.8, "label": "Коэффициент φₖ"},
            annot_kws={"size": 10, "fontweight": "bold"})
    
plt.title('Матрица корреляции признаков заведений\n(коэффициент φₖ)', fontsize=16, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.show()
```

```python
 # Получаем корреляции с рейтингом
rating_phik = phik_overview['rating'].sort_values(ascending=False)
    
print("\nКОРРЕЛЯЦИЯ ПРИЗНАКОВ С РЕЙТИНГОМ:")
print("="*60)
    
# Интерпретация силы корреляции
def interpret_phik_strength(value):
    if value >= 0.3:
        return "ОЧЕНЬ СИЛЬНАЯ"
    elif value >= 0.2:
        return "СИЛЬНАЯ"
    elif value >= 0.1:
        return "УМЕРЕННАЯ"
    elif value >= 0.05:
        return "СЛАБАЯ"
    else:
        return "ОЧЕНЬ СЛАБАЯ"
    
for feature, phi_value in rating_phik.items():
    if feature != 'rating':
        strength = interpret_phik_strength(abs(phi_value))
        direction = "положительная" if phi_value > 0 else "отрицательная"
        print(f"{feature:20}: {phi_value:7.3f} | {strength:15} {direction} корреляция")
```

```python
#АНАЛИЗ САМОЙ СИЛЬНОЙ СВЯЗИ
print("\nАНАЛИЗ САМОЙ СИЛЬНОЙ СВЯЗИ С РЕЙТИНГОМ:")
    
# Находим признак с самой сильной корреляцией (исключая сам рейтинг)
strongest_features = rating_phik.drop('rating')
strongest_feature = strongest_features.abs().idxmax()
strongest_value = strongest_features[strongest_feature]
    
print(f"Самый сильный признак: {strongest_feature}")
print(f"Коэффициент φₖ с рейтингом: {strongest_value:.3f}")
print(f"Сила связи: {interpret_phik_strength(abs(strongest_value))}")
print(f"Направление: {'положительная' if strongest_value > 0 else 'отрицательная'}")
```

```python
# ДЕТАЛИЗАЦИЯ САМОЙ СИЛЬНОЙ СВЯЗИ
 
# Создаем табличный анализ в зависимости от типа признака
if strongest_feature in ['category', 'district', 'chain', 'is_24_7']:
    # Для категориальных признаков: группируем и считаем статистику
    group_stats = phi_data.groupby(strongest_feature)['rating'].agg([
        'count', 'mean', 'median', 'std', 'min', 'max'
    ]).round(3)
        
# Переименовываем столбцы для читаемости
    group_stats.columns = ['Кол-во заведений', 'Средний рейтинг', 'Медиана рейтинга', 
                               'Станд. отклонение', 'Минимальный рейтинг', 'Максимальный рейтинг']
        
    # Сортируем по среднему рейтингу
    group_stats = group_stats.sort_values('Средний рейтинг', ascending=False)
        
    print(f"\nСтатистика рейтинга по группам '{strongest_feature}':")
    print(group_stats.to_string())
        
elif strongest_feature in ['seats', 'price']:
    # Для числовых признаков: создаем интервальные группы
    if strongest_feature == 'seats':
        # Для количества мест создаем логичные интервалы
        phi_data[f'{strongest_feature}_group'] = pd.cut(
            phi_data[strongest_feature], 
            bins=[0, 10, 30, 50, 100, 500],
            labels=['1-10 мест', '11-30 мест', '31-50 мест', '51-100 мест', '101-500 мест']
        )
    else:
        # Для price (категориальный, но преобразованный в числовой)
        # price уже должен быть категориальным
        phi_data[f'{strongest_feature}_group'] = phi_data[strongest_feature]
    
    # Группируем по созданным группам
    group_stats = phi_data.groupby(f'{strongest_feature}_group')['rating'].agg([
        'count', 'mean', 'median', 'std', 'min', 'max'
    ]).round(3)
        
    group_stats.columns = ['Кол-во заведений', 'Средний рейтинг', 'Медиана рейтинга', 
                               'Станд. отклонение', 'Минимальный рейтинг', 'Максимальный рейтинг']
        
    group_stats = group_stats.sort_values('Средний рейтинг', ascending=False)
        
    print(f"\nСтатистика рейтинга по группам '{strongest_feature}':")
    print(group_stats.to_string())
```

```python
# ВИЗУАЛИЗАЦИЯ САМОЙ СИЛЬНОЙ СВЯЗИ
print(f"ВИЗУАЛИЗАЦИЯ СВЯЗИ РЕЙТИНГА:")

# Создаем визуализацию в зависимости от типа признака
if strongest_feature in ['category', 'district', 'chain', 'is_24_7', 'price']:
    # Для категориальных признаков: столбчатая диаграмма

    # Готовим данные для визуализации
    if strongest_feature in ['seats', 'price'] and f'{strongest_feature}_group' in phi_data.columns:
        # Используем созданные группы для seats и price
        viz_data = phi_data.groupby(f'{strongest_feature}_group')['rating'].mean().sort_values(ascending=False)
        x_labels = [str(label) for label in viz_data.index]
    else:
        # Для обычных категориальных признаков
        viz_data = phi_data.groupby(strongest_feature)['rating'].mean().sort_values(ascending=False)
        x_labels = [str(label)[:20] + '...' if len(str(label)) > 20 else str(label) 
                   for label in viz_data.index]

    # Берем топ-15 категорий для наглядности
    top_n = min(15, len(viz_data))
    top_categories = viz_data.head(top_n)

    plt.figure(figsize=(14, 8))

    # Создаем столбчатую диаграмму
    bars = plt.bar(range(len(top_categories)), top_categories.values, 
                  color=plt.cm.viridis(np.linspace(0, 1, len(top_categories))),
                  edgecolor='black', alpha=0.8)

    plt.xlabel(strongest_feature, fontsize=12)
    plt.ylabel('Средний рейтинг', fontsize=12)
    plt.title(f'Средний рейтинг заведений по признаку "{strongest_feature}"\n(коэффициент φₖ = {strongest_value:.3f})', 
              fontsize=14, fontweight='bold')
    plt.xticks(range(len(top_categories)), 
              [str(label)[:15] + '...' if len(str(label)) > 15 else str(label) 
               for label in top_categories.index],
              rotation=45, ha='right', fontsize=10)
    plt.grid(axis='y', alpha=0.3)

    # Добавляем линию общего среднего рейтинга
    overall_mean = phi_data['rating'].mean()
    plt.axhline(y=overall_mean, color='red', linestyle='--', linewidth=2, 
               label=f'Общий средний рейтинг: {overall_mean:.2f}')

    # Добавляем значения на столбцы
    for i, value in enumerate(top_categories.values):
        plt.text(i, value + 0.01, f"{value:.2f}", 
                ha='center', va='bottom', fontsize=9, fontweight='bold')

    plt.legend(fontsize=11)
    plt.tight_layout()
    plt.show()

elif strongest_feature == 'seats':
    # Для числового признака seats: scatter plot с линией тренда
    plt.figure(figsize=(14, 8))

    # Создаем scatter plot
    plt.scatter(phi_data['seats'], phi_data['rating'], 
               alpha=0.5, s=20, color='steelblue')

    plt.xlabel('Количество посадочных мест', fontsize=12)
    plt.ylabel('Рейтинг заведения', fontsize=12)
    plt.title(f'Зависимость рейтинга от количества мест\n(коэффициент φₖ = {strongest_value:.3f})', 
              fontsize=14, fontweight='bold')
    plt.grid(alpha=0.3)

    # Добавляем линию тренда (линейная регрессия)
    # Исключаем пропуски для расчета тренда
    valid_data = phi_data[['seats', 'rating']].dropna()
    if len(valid_data) > 1:
        x = valid_data['seats']
        y = valid_data['rating']

        # Линейная регрессия
        z = np.polyfit(x, y, 1)
        p = np.poly1d(z)

        # Создаем диапазон для линии тренда
        x_trend = np.linspace(x.min(), x.max(), 100)
        plt.plot(x_trend, p(x_trend), "r-", linewidth=2, 
                label=f'Линия тренда: y = {z[0]:.4f}x + {z[1]:.2f}')

    plt.legend(fontsize=11)
    plt.tight_layout()
    plt.show()

    # Дополнительно: box plot по группам мест
    plt.figure(figsize=(14, 8))

    # Создаем группы мест
    phi_data['seats_group'] = pd.cut(
        phi_data['seats'], 
        bins=[0, 10, 30, 50, 100, 500],
        labels=['1-10 мест', '11-30 мест', '31-50 мест', '51-100 мест', '101-500 мест']
    )

    # Box plot по группам
    box_data = []
    box_labels = []
    for group in ['1-10 мест', '11-30 мест', '31-50 мест', '51-100 мест', '101-500 мест']:
        group_ratings = phi_data[phi_data['seats_group'] == group]['rating']
        if len(group_ratings) > 0:
            box_data.append(group_ratings)
            box_labels.append(group)

    bp = plt.boxplot(box_data, labels=box_labels, patch_artist=True)

    # Раскрашиваем боксы
    colors = ['lightcoral', 'lightgreen', 'lightblue', 'gold', 'lightpink']
    for patch, color in zip(bp['boxes'], colors):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    plt.xlabel('Группы по количеству мест', fontsize=12)
    plt.ylabel('Рейтинг заведения', fontsize=12)
    plt.title('Распределение рейтинга по группам количества мест', fontsize=14, fontweight='bold')
    plt.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    plt.show()
```

На основе корреляционного анализа с использованием универсального коэффициента φₖ выявлено, что наиболее сильная связь рейтинга заведений наблюдается с ценовой категорией. Коэффициент φₖ составляет 0.220, что классифицируется как сильная положительная корреляция. Это означает, что заведения более высокой ценовой категории статистически значимо чаще получают более высокие оценки: рестораны высокой ценовой категории имеют средний рейтинг 4.44, тогда как заведения с низкими ценами — 4.17. Данная закономерность указывает на то, что ценовой сегмент является важным фактором, формирующим ожидания и восприятие качества у посетителей.

При этом анализ показал, что административный район и категория заведения также демонстрируют заметную связь с рейтингом (φₖ = 0.201 и 0.190 соответственно), однако она уступает влиянию цены. Интересно, что такие формальные характеристики, как количество посадочных мест, статус сетевого заведения и круглосуточный режим работы, демонстрируют очень слабую или умеренную связь с рейтингом (φₖ ≤ 0.150). Это позволяет сделать вывод, что для повышения рейтинга владельцам в первую очередь стоит фокусироваться на качестве предоставляемых услуг и создании уникального опыта, соответствующего ценовому позиционированию, а не только на увеличении площади или формальных атрибутах заведения.

```python
# Средний чек по категориям в разных типах районов
plt.figure(figsize=(14, 8))

# Готовим данные для группированного bar chart
comparison_data = []
for category in top_categories:
    # ЦАО
    cao_vals = cao_data[cao_data['category'] == category]['middle_avg_bill']
    cao_mean = cao_vals.mean() if len(cao_vals) > 0 else 0
    
    # Центральные (без ЦАО)
    central_non_cao = central_data[(central_data['category'] == category) & 
                                  (central_data['district'] != 'Центральный административный округ')]['middle_avg_bill']
    central_mean = central_non_cao.mean() if len(central_non_cao) > 0 else 0
    
    # Периферийные
    peripheral_vals = peripheral_data[peripheral_data['category'] == category]['middle_avg_bill']
    peripheral_mean = peripheral_vals.mean() if len(peripheral_vals) > 0 else 0
    
    comparison_data.append({
        'Категория': category,
        'ЦАО': cao_mean,
        'Др. центральные': central_mean,
        'Периферийные': peripheral_mean
    })

comparison_df = pd.DataFrame(comparison_data).set_index('Категория')
```

```python
# Создаем группированный bar chart
x = np.arange(len(comparison_df))
width = 0.25

fig, ax = plt.subplots(figsize=(14, 8))
bars1 = ax.bar(x - width, comparison_df['ЦАО'], width, label='ЦАО')
bars2 = ax.bar(x, comparison_df['Др. центральные'], width, label='Другие центральные')
bars3 = ax.bar(x + width, comparison_df['Периферийные'], width, label='Периферийные')

ax.set_xlabel('Категория заведения', fontsize=12)
ax.set_ylabel('Средний чек, руб.', fontsize=12)
ax.set_title('Средний чек по категориям заведений в разных типах районов', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(comparison_df.index, rotation=45, ha='right')
ax.legend(title='Тип района')
ax.grid(axis='y', alpha=0.3)

# Добавляем значения на столбцы
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax.text(bar.get_x() + bar.get_width()/2, height + 20,
                    f'{height:.0f}', ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()
```

### **3. Рекомендации для заказчика**

**Для новых игроков:**
1. **Рассмотреть нишевые форматы** - гибридные концепции (бар+паб) показывают устойчивость
2. **Оценить периферийные районы** - менее конкурентная среда, потенциал роста
3. **Сфокусироваться на качестве** - высокие рейтинги несетевых заведений указывают на важность уникальности

**Для существующих операторов:**
1. **Оптимизировать локации** - учитывать ценовой градиент "центр-периферия"
2. **Дифференцировать ценовую политику** в зависимости от района расположения
3. **Развивать форматы с 0 посадочных мест** - перспективное направление доставки

1. **Акцентировать локальность** - несетевые заведения имеют конкурентное преимущество в рейтингах
2. **Использовать географический фактор** - позиционирование в ЦАО как премиум-локации
3. **Разработать дифференцированные предложения** для разных типов районов

**Операционные рекомендации**

1. **Адаптировать вместимость** под формат и локацию:
   - Центр: средние и крупные форматы (40-80 мест)
   - Периферия: микро-форматы и точки доставки
   
2. **Оптимизировать график работы**:
   - Только 0.5% заведений работают круглосуточно
   - Потенциал для развития ночных форматов в ЦАО

3. **Мониторить качество обслуживания**:
   - Высокие базовые ожидания потребителей (средний рейтинг 4.4)
   - Незначительные преимущества дают конкурентное преимущество

**Инвестиционные рекомендации**

1. **Приоритетные сегменты для инвестиций**:
   - **Кофейни** - высокие рейтинги, растущая сетевизация
   - **Нишевые гибридные форматы** - низкая конкуренция
   - **Форматы доставки** - 43% рынка с 0 посадочных мест

2. **Географические приоритеты**:
   - **ЦАО** для премиум-сегмента
   - **Периферийные районы** для демократичных форматов
   - **Новые районы** с низкой насыщенностью

3. **Форматные рекомендации**:
   - **Сетевые концепции** - для фастфуда и кофеен
   - **Уникальные локальные проекты** - для ресторанов и столовых
